# Chapter 1 · What is seismic 4C?

## Learning goal
Understand what the four components measure, how a typical marine 4C receiver is arranged, and why the measurements complement one another.

## Brief theory
**4C** means four-component seismic recording. In a common ocean-bottom setup, one receiver package contains:

| Component | Typical sensor | What it measures |
|---|---|---|
| **P** | Hydrophone | Acoustic pressure in the water |
| **Z** | Vertical geophone | Vertical seabed particle motion |
| **X** | Horizontal geophone | Seabed particle motion along one horizontal axis |
| **Y** | Horizontal geophone | Seabed particle motion along the perpendicular horizontal axis |

The hydrophone measures **pressure**, while the three geophones measure **ground motion** (commonly particle velocity, depending on sensor design). The extra motion components can help capture converted shear-wave energy and provide information complementary to pressure-only data. Joint use can improve imaging and characterization, but the benefit depends on geology, acquisition, coupling, and processing.

> **Keep in mind:** “4C” describes four recorded channels, not four wave types. The diagram and signals below are simplified teaching illustrations, not field data.

## Visual: a typical ocean-bottom 4C receiver

A receiver rests on the seabed. A nearby source sends energy into the subsurface; reflected energy returns to the receiver and is recorded on all four channels. The drawing is schematic and not to scale.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Circle, FancyArrowPatch

fig, ax = plt.subplots(figsize=(10, 5))
ax.set_xlim(0, 10)
ax.set_ylim(0, 6)
ax.set_aspect('equal')
ax.axis('off')

# Water column, seabed, and a simple subsurface reflector
ax.add_patch(Rectangle((0, 2.2), 10, 3.8, color='#dff3ff'))
ax.add_patch(Rectangle((0, 0), 10, 2.2, color='#e8d7b5'))
ax.plot([0, 10], [2.2, 2.2], color='#356b7a', lw=2)
ax.plot([1, 9], [0.8, 0.8], color='#956f3a', lw=2, ls='--')
ax.text(0.3, 5.55, 'Water', color='#24556b', fontsize=12)
ax.text(0.3, 1.8, 'Seabed', color='#684c2b', fontsize=11)
ax.text(7.1, 0.95, 'Subsurface reflector', color='#684c2b', fontsize=10)

# Source and receiver package
ax.add_patch(Circle((7.9, 4.8), 0.16, color='#e4572e'))
ax.text(8.15, 4.72, 'Marine source', color='#8c321e', fontsize=10)
ax.add_patch(Rectangle((3.75, 2.22), 1.7, 0.22, color='#444444'))
ax.text(3.55, 2.56, 'Ocean-bottom 4C receiver', fontsize=10, weight='bold')

# Four channels in the receiver package
components = [('P', 'hydrophone'), ('Z', 'vertical'), ('X', 'horizontal'), ('Y', 'horizontal')]
for i, (name, label) in enumerate(components):
    x = 3.95 + i * 0.42
    ax.add_patch(Circle((x, 2.33), 0.09, color=['#277da1', '#43aa8b', '#f9c74f', '#9b5de5'][i]))
    ax.text(x, 1.95, f'{name}\n{label}', ha='center', va='top', fontsize=9)

# Illustrative ray path
path = [(7.9, 4.65), (6.6, 3.45), (5.0, 0.85), (4.6, 2.2)]
for start, end in zip(path[:-1], path[1:]):
    ax.add_patch(FancyArrowPatch(start, end, arrowstyle='->', mutation_scale=12,
                                 lw=1.6, color='#d1495b', connectionstyle='arc3,rad=0.08'))
ax.text(6.15, 3.0, 'reflected energy', color='#a83245', fontsize=9, rotation=-48)
ax.set_title('A simplified marine ocean-bottom 4C survey', fontsize=15, pad=12)
plt.show()
```

## Practical example: compare four synthetic traces

Below, we create one simple wavelet and scale it differently for each component. This demonstrates that the channels can have different amplitudes and units; it is **not** a physical simulation of 4C wave propagation. Run the cell and compare the four traces.

```python
import numpy as np

# A short Ricker-like pulse centered at 0.35 s
t = np.linspace(0, 0.8, 1200)
f_peak = 18.0  # Hz
t0 = 0.35      # s
u = np.pi * f_peak * (t - t0)
pulse = (1 - 2 * u**2) * np.exp(-u**2)

# Illustrative component amplitudes and small time shifts
traces = {
    'P · pressure': pulse,
    'Z · vertical motion': 0.75 * np.interp(t - 0.008, t, pulse, left=0, right=0),
    'X · horizontal motion': 0.45 * np.interp(t - 0.015, t, pulse, left=0, right=0),
    'Y · horizontal motion': -0.30 * np.interp(t - 0.020, t, pulse, left=0, right=0),
}

fig, axes = plt.subplots(4, 1, figsize=(10, 7), sharex=True)
for ax, (label, trace) in zip(axes, traces.items()):
    ax.plot(t, trace, lw=1.4)
    ax.axhline(0, color='0.7', lw=0.7)
    ax.set_ylabel(label, rotation=0, ha='right', va='center', labelpad=12)
    ax.grid(alpha=0.2)
axes[-1].set_xlabel('Time (s)')
fig.suptitle('Illustrative 4C traces — synthetic, not field data', y=1.01)
fig.tight_layout()
plt.show()
```

### What to notice
- All channels may respond to the same event, but their waveforms need not match.
- Pressure and motion are different measurements, so their amplitudes and units are not directly interchangeable.
- The two horizontal components preserve motion in two directions. Orientation and coupling matter when interpreting them.

## One-sentence summary
**Marine 4C seismic records pressure plus three orthogonal components of seabed motion, giving complementary observations of the subsurface.**
